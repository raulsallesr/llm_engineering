```python
import os
from dotenv import load_dotenv
from openai import OpenAI # biblioteca OpenIa
from scraper import fetch_website_contents
```

```python
# Carregar as variaveis do .env
load_dotenv()

# key open ai
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
MODEL = "gpt-4.1-mini"
```

```python
#site
url = "https://ge.globo.com/"

# conteúdo do site
content = fetch_website_contents(url)
print("Conteúdo Coletado: ")
print(content[:500])
```

```python
# Personalidade para IA
system_prompt = """
Você é um assistente que cria resumos curtos de notícias esportivas.
"""
# Dados p IA
# somente os primeiros 5000 caracteres
user_prompt = f"""
Resuma as notícias mais importantes desse conteúdo:

{content[:5000]} 
"""
```

```python
# Chamando OPENAI
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role":"system", "content": system_prompt},
        {"role":"user","content":user_prompt}
    ]
)
```

```python
summary = response.choices[0].message.content # pegar a primeira resposta

print("\nResumo Final:\n")
print(summary)
```
